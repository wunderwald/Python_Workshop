# Cheat Sheet: Python Basics

All the syntax from day 1 in one place. Every code cell can be run, and the expected output is written next to each line as a comment (`# -> ...`).

### Comments and print()

````python
# this is a comment

'''
this is a
multi-line comment
'''

print("hello world")         # -> hello world
print("a", "b", 3)           # -> a b 3
print("same line", end=" ")  # end=" " prints WITHOUT going to a new line
print("!")                   # -> same line !
print()                      # prints only a new line
````

### Variables and types

- Variables are created with `=`: `variable_name = value`
- Python determines the type automatically. `type()` tells you what it is.

````python
my_int = -10
my_float = 3.12
my_bool = True
my_string = "hello"

print(type(my_int))     # -> <class 'int'>
print(type(my_float))   # -> <class 'float'>
print(type(my_bool))    # -> <class 'bool'>
print(type(my_string))  # -> <class 'str'>
````

### Strings

````python
single = 'single quotes'
double = "double quotes"
mixed = "use 'single quotes' inside double quotes"
escaped = "or escape them with a backslash: \"like this\""
multiline = "line one\nline two"    # \n creates a new line
print(escaped)                      # -> or escape them with a backslash: "like this"
print(multiline)                    # -> line one
                                    #    line two

# adding strings
greeting = "hello" + " " + "world"
print(greeting)                     # -> hello world

# useful string functions and methods
print(len(greeting))                # -> 11
print(greeting.upper())             # -> HELLO WORLD
print(greeting.startswith("he"))    # -> True
````

**f-strings** put variables and calculations into strings: `f"text {variable}"`

````python
fav = 333
print(f"My fav number is {fav}")         # -> My fav number is 333
print(f"Twice my fav number is {fav * 2}")  # -> Twice my fav number is 666
````

### Math

````python
a = 6
b = 4

print(a + b)    # -> 10
print(a - b)    # -> 2
print(a * b)    # -> 24
print(a / b)    # -> 1.5   (division always returns a float)
print(a // b)   # -> 1     (floor division: rounded down)
print(a % b)    # -> 2     (modulo: the remainder of a division)
print(a ** 2)   # -> 36    (power)

a += 1          # short for: a = a + 1
print(a)        # -> 7

print(round(3.14159, 2))   # -> 3.14
````

### Casting

- **Explicit casting**: convert a value yourself with `int()`, `float()`, `str()`, `bool()`
- **Implicit casting**: Python converts types automatically during some operations

````python
# explicit
print(int(7.777))     # -> 7      (cuts off the decimals)
print(float(3))       # -> 3.0
print(str(9.99))      # -> 9.99   (now a string)
print(float("-4"))    # -> -4.0
print(int(True))      # -> 1      (True is 1, False is 0)
print(bool(0))        # -> False  (0 and "" are "falsy")
print(bool("hi"))     # -> True   (most other values are "truthy")

# implicit
print(2 + 3.5)        # -> 5.5    (int + float => float)
print(True + 8)       # -> 9      (bool + int => int)
print(4 / 2)          # -> 2.0    (division => float)

# not always possible:
# "40" + 40  -> TypeError! Cast explicitly instead:
print(int("40") + 40)   # -> 80
print("40" + str(40))   # -> 4040
````

### Boolean logic and comparisons

````python
a = True
b = False

print(a and b)        # -> False  (both must be True)
print(a or b)         # -> True   (at least one must be True)
print(not a)          # -> False
print(not (a and b))  # -> True   (brackets build more complex conditions)

x = 5
print(x == 5)   # -> True   (equals)
print(x != 5)   # -> False  (does not equal)
print(x < 3)    # -> False  (less than)
print(x <= 5)   # -> True   (less than or equal)
print(x > 3)    # -> True   (greater than)
print(x >= 6)   # -> False  (greater than or equal)
````

### Conditionals

The code inside each branch must be **indented**. Only the first branch whose condition is True runs.

````python
stress = 4

if stress >= 6:
    print("high stress")
elif stress >= 3:
    print("medium stress")
else:
    print("low stress")
# -> medium stress
````

### Lists

Lists are **ordered** and **mutable** (can be changed). `[ ... ]`

````python
my_list = [3, 1, 2]

my_list.append(5)          # add to the end
print(my_list)             # -> [3, 1, 2, 5]

my_list.remove(1)          # remove the first 1 (error if it's not there)
print(my_list)             # -> [3, 2, 5]

last = my_list.pop()       # remove and return the last item (pop(i): the item at i)
print(last, my_list)       # -> 5 [3, 2]

print(my_list.count(3))    # -> 1  (how often 3 appears)
print(len(my_list))        # -> 2  (number of elements)

my_list.sort()             # sorts the list itself ("in place")
print(my_list)             # -> [2, 3]
my_list.reverse()          # reverses the list itself
print(my_list)             # -> [3, 2]
````

**Indexing and slicing** (works for lists, tuples and strings). Counting starts at **0**!

````python
letters = ["a", "b", "c", "d", "e"]

print(letters[0])     # -> a              (first element)
print(letters[-1])    # -> e              (negative: count from the end)
print(letters[1:3])   # -> ['b', 'c']     (from 1 up to 3, 3 is excluded)
print(letters[:2])    # -> ['a', 'b']     (from the start)
print(letters[2:])    # -> ['c', 'd', 'e']  (until the end)

letters[0] = "z"      # change an element
print(letters)        # -> ['z', 'b', 'c', 'd', 'e']

print("python"[0:2])  # -> py  (strings can be sliced too)
````

**Matrix**: a list of lists

````python
matrix = [
    [0, 1, 2],
    [3, 4, 5],
    [6, 7, 8],
]

print(matrix[1])      # -> [3, 4, 5]  (row 1)
print(matrix[1][2])   # -> 5          (row 1, column 2)
````

### Tuples and sets

- **Tuple** `( ... )`: ordered, **immutable** (can't be changed after creation)
- **Set** `{ ... }`: unordered, mutable, every element is **unique**

````python
my_tuple = (4, 5, 6, 4)
print(my_tuple.count(4))   # -> 2
print(my_tuple.index(6))   # -> 2  (position of 6)
# my_tuple.append(7)       -> AttributeError: tuples can't be changed

my_set = {"cat", "dog"}
my_set.add("bird")         # add one item
my_set.add("cat")          # nothing happens, "cat" is already in the set
my_set.remove("dog")       # remove an item (error if it's not there)
print(len(my_set))         # -> 2
````

**Casting between them** with `list()`, `tuple()`, `set()`. This can change the data!

````python
locations = ["home", "work", "home", "park"]

print(tuple(locations))      # -> ('home', 'work', 'home', 'park')
print(set(locations))        # -> {'home', 'work', 'park'}  (duplicates removed, order may vary)
print(len(set(locations)))   # -> 3  (number of different values)
````

### Dicts

Dicts store **key-value pairs**. Keys are usually strings, values can be anything (even lists or other dicts).

````python
participant = {
    "id": "P017",
    "age": 24,
    "languages": ["German", "English"],
    "sleep_hours": {"Mon": 7.0, "Tue": 5.5},
}

print(participant["age"])                 # -> 24       (lookup)
print(participant["sleep_hours"]["Tue"])  # -> 5.5      (nested lookup)
print(participant["languages"][0])        # -> German   (lookup + list index)

participant["age"] = 25                   # change a value
participant["excluded"] = False           # add a new key-value pair

print(participant.keys())    # -> dict_keys(['id', 'age', 'languages', 'sleep_hours', 'excluded'])
print("age" in participant)  # -> True   (does this key exist?)
print("name" in participant) # -> False  (participant["name"] would be a KeyError)
````

### For loops

A for loop runs once **for each value** in a list (or tuple, set, string, dict...).

````python
for letter in ["a", "b", "c"]:
    print(letter, end=" ")      # -> a b c
print()

for i in range(3):              # range(3) => 0, 1, 2
    print(i, end=" ")           # -> 0 1 2
print()

for i in range(4, 7):           # range(4, 7) => 4, 5, 6 (7 is excluded)
    print(i, end=" ")           # -> 4 5 6
print()

sleep_hours = {"Mon": 7.0, "Tue": 5.5}
for day in sleep_hours:         # looping over a dict gives you its keys
    print(day, sleep_hours[day])  # -> Mon 7.0
                                  #    Tue 5.5
````

Two common loop patterns: **adding up** and **collecting**.

````python
values = [3, 5, 4, 6]

total = 0                  # start BEFORE the loop
for value in values:
    total += value
print(total / len(values))  # -> 4.5

high = []                  # start with an empty list
for value in values:
    if value >= 5:
        high.append(value)
print(high)                # -> [5, 6]
````

### While loops

A while loop runs **as long as** its condition is True:

- check the condition
- if it is False, stop
- if it is True, run the code inside and repeat

Make sure the condition becomes False at some point! A loop that never ends can be stopped with the ■ (interrupt) button.

````python
count = 0
while count < 3:
    print(count, end=" ")   # -> 0 1 2
    count += 1
print()
````

### Functions

Reusable pieces of code with inputs (**arguments**) and outputs (**return values**).

````python
def calc_mean(values):
    total = 0
    for value in values:
        total += value
    return total / len(values)

result = calc_mean([3, 5, 4])
print(result)                # -> 4.0


# arguments can have default values
def greet(name, language="en"):
    if language == "de":
        return f"Hallo {name}!"
    return f"Hello {name}!"

print(greet("Ana"))                 # -> Hello Ana!
print(greet("Ana", language="de"))  # -> Hallo Ana!
````

**print vs. return**: `print()` only shows a value. `return` hands it back so you can store it in a variable and keep working with it.

### Packages

Packages are collections of modules written by other people.

- The **standard library** comes with Python (e.g. `math`, `random`, `statistics`, `datetime`, `json`)
- **Third-party packages** are installed in the terminal: `pip install <package-name>`
- `help(package)` or `dir(package)` show what's inside, or search for the documentation online
- Never name your own files like a package (e.g. `random.py`)!

````python
import math
import random
import statistics
from datetime import date

print(math.sqrt(16))                  # -> 4.0
print(statistics.mean([3, 5, 4]))     # -> 4
print(random.randint(1, 7))           # -> a random integer from 1 to 7
print(date.today())                   # -> today's date
````

### Tips & tricks

````python
# list comprehension: a for loop inside a list
squares = [x ** 2 for x in [1, 2, 3]]
print(squares)                             # -> [1, 4, 9]

# ... with a filter
print([x for x in [1, 5, 2, 8] if x > 2])  # -> [5, 8]

# ternary operator: if/else in one line
formal = True
greeting = "Kind regards" if formal else "xoxo"
print(greeting)                            # -> Kind regards

# zip(): loop over two lists at once
for name, age in zip(["Ana", "Ben"], [31, 27]):
    print(f"{name}: {age}")                # -> Ana: 31
                                           #    Ben: 27

# enumerate(): index and value
for i, day in enumerate(["Mon", "Tue"]):
    print(i, day)                          # -> 0 Mon
                                           #    1 Tue
````

### Errors

Errors and handling are not covered in this workshop but here's the basics.

#### Reading error messages

Read error messages **from the bottom**: the last line says **what** went wrong, the line number above says **where**.

| Error | Typical cause | Example |
|---|---|---|
| `SyntaxError` | Typo in the code structure (missing `:`, bracket or quote) | `if x > 3` |
| `IndentationError` | Wrong or missing indentation | code inside `if` not indented |
| `NameError` | Variable or function doesn't exist (yet), or is misspelled | `print(undefined_variable)` |
| `TypeError` | Operation doesn't work for these types | `"40" + 40` |
| `ValueError` | Right type, but impossible value | `int("seven")` |
| `IndexError` | Index is too large for the list | `[1, 2][5]` |
| `KeyError` | Key doesn't exist in the dict | `{"a": 1}["b"]` |
| `AttributeError` | The object doesn't have this method | `(1, 2).append(3)` |
| `ZeroDivisionError` | Division by zero | `10 / 0` |

#### Catching errors

Normally an error stops your code. With `try` / `except`, you can **catch** an error and decide what should happen instead.

```python
try:
    ... code that might fail ...
except SomeError:
    ... what to do if it fails ...
```

- Always name the error you expect (e.g. `except ValueError:`). A bare `except:` also hides errors you didn't expect, which makes bugs hard to find.
- `except ... as e` gives you the error message.

````python
# catch one specific error
try:
    number = int("seven")
except ValueError:
    print("That's not a number!")       # -> That's not a number!

# handle different errors differently
participant = {"id": "P017", "age": 24}
try:
    print(participant["name"])
except KeyError as e:
    print(f"Missing key: {e}")          # -> Missing key: 'name'

# typical use: skip bad values in a loop instead of crashing
answers = ["5", "7", "seven", "3"]
total = 0
for answer in answers:
    try:
        total += int(answer)
    except ValueError:
        print(f"Skipping '{answer}': not a number")   # -> Skipping 'seven': not a number
print(total)                            # -> 15
````
