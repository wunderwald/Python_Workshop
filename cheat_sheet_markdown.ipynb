# Cheat Sheet: Markdown

Markdown cells contain formatted text. **Double-click** a cell to see its source, press **Shift + Enter** to render it again.

### Headlines

```markdown
# Headline 1
## Headline 2
### Headline 3
```

The more `#`, the smaller the headline. Don't forget the space after the `#`!

### Text formatting

```markdown
**bold**, *italic*, ~~strikethrough~~

An empty line starts a new paragraph.
```

**Result:**

**bold**, *italic*, ~~strikethrough~~

An empty line starts a new paragraph.

### Links (URLs)

```markdown
[Python documentation](https://docs.python.org)

<https://pypi.org>
```

**Result:**

[Python documentation](https://docs.python.org)

<https://pypi.org>

### Code

````markdown
Inline code: `print("hello")`

A code block:
```python
for i in range(3):
    print(i)
```
````

**Result:**

Inline code: `print("hello")`

A code block:
```python
for i in range(3):
    print(i)
```

### Lists

```markdown
- bullet point
- another one
    - indented with 4 spaces or tab

1. first
2. second
3. third
```

**Result:**

- bullet point
- another one
    - indented with 4 spaces or tab

1. first
2. second
3. third

### Tables

```markdown
| Name | Age |
|------|-----|
| Ana  | 31  |
| Ben  | 27  |
```

**Result:**

| Name | Age |
|------|-----|
| Ana  | 31  |
| Ben  | 27  |
