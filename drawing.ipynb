# Drawing with code

A small Processing-style toolkit: you call functions like `rect()` or `circle()`, and a picture appears below the cell.

* `(0, 0)` is the **top-left** corner. `x` grows to the right, `y` grows **downwards**.
* Start every new picture with `size(width, height)`.
* `fill()` sets the colour *inside* shapes, `stroke()` the colour of outlines and lines. They apply to everything drawn after them.

Run the next cell once to load the toolkit.

```python
import os
import urllib.request

# In Colab, download sketch.py first (locally it is already next to this notebook)
if not os.path.exists("sketch.py"):
    url = "https://raw.githubusercontent.com/wunderwald/Python_Workshop/main/sketch.py"
    urllib.request.urlretrieve(url, "sketch.py")

from sketch import *
```

## A first picture

```python
size(400, 300)
background("white")

fill(255, 200, 0)       # red, green, blue
circle(200, 150, 120)   # x, y of the centre, diameter

stroke("crimson")
stroke_weight(4)
line(0, 0, 400, 300)    # from (0, 0) to (400, 300)

no_stroke()
fill(0, 120, 255, 120)  # the 4th number makes it see-through
rect(250, 50, 120, 200) # x, y of the top-left corner, width, height
```

## One loop: a row of circles

The loop variable `i` counts 0, 1, 2, ... and we use it to move each circle further to the right.

```python
size(500, 100)
background(30)

for i in range(10):
    circle(25 + i * 50, 50, 40)
```

## Colour from the loop

`hsb(hue)` picks a colour from the colour wheel: hue 0 is red, 120 green, 240 blue, 360 red again.

```python
size(500, 100)
background(30)
no_stroke()

for i in range(10):
    fill(hsb(i * 36))
    circle(25 + i * 50, 50, 40)
```

## Two loops inside each other: a grid

```python
size(400, 400)
background("white")
no_stroke()

for row in range(10):
    for col in range(10):
        fill(col * 25, row * 25, 150)   # colour depends on the position
        square(col * 40, row * 40, 38)
```

## Randomness

```python
import random

size(400, 400)
background(20)
no_stroke()

for i in range(300):
    x = random.uniform(0, 400)
    y = random.uniform(0, 400)
    d = random.uniform(5, 40)
    fill(hsb(random.uniform(180, 300), 70, 100), 150)
    circle(x, y, d)
```
