```python
from eeg_simulation import simulate_eeg
from plot import lineplot

trials, channels, times = simulate_eeg(n_trials=1)
eeg_ch0 = np.array(trials[0][0])

lineplot(eeg_ch0)

# dc correction in python:
# calculate mean
sum = 0
num_samples = len(eeg_ch0)
for sample in eeg_ch0:
    sum += sample
mean = sum / num_samples

# substract from all samples
corrected_ch0 = []
for sample in eeg_ch0:
    corrected_sample = sample - mean
    corrected_ch0.append(corrected_sample)

lineplot(corrected_ch0)
```
