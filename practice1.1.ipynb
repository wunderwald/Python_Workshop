### Casting

To change the type of a primitive to another, use explicit casting functions:

- int()
- float()
- bool()
- str()

3 Tasks ⬇

````python
# 1. Play with casting: create a float, convert it to an integer, then a string, then a boolean.
# ---> how does the value change at each step?

f = 1.111
i = int(f)
s = str(i)
b = bool(s)

print(f"{f} ===> {i} ===> {s} ===> {b}")
# 1.111 ===> 1 ===> 1 ===> True

# 2. The bool() function determines if a value is "truthy" or "falsy".
# ---> find out: which int(s) and "string(s)" are True, which are False?
print(bool(0))
print(bool(1))
print(bool(""))
print(bool("something"))
# False
# True
# False
# True

# 3. use some casting magic to make 6 + 7 == 67
six = 6
seven = 7
result = int(str(six) + str(seven))
print(f"6 + 7 = {result}")
````
