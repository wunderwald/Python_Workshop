### Array-Likes

There are three types that store multiple objects in python:

- list ```[ ... ]```: ordered, mutable
- tuple ```( ... )```: ordered, **immutable**
- set ```{ ... }```: unordered, mutable, **unique elements**

#### Important methods

**List**:

- ```my_list.append(x)```: add x to the end
- ```my_list.remove(x)```: remove the first x (error if x isn't there)
- ```my_list.pop() / my_list.pop(i)	```: remove and return the last item / the item at i
- ```my_list.count(x)```: count how often x appears
- ```my_list.sort(), my_list.reverse()```: sort / reverse the list (*in place*)

**Tuple**:

- ```my_tuple.count(x)```: how often x appears
- ```my_tuple.index(x)```

**Set**

- ```my_set.add(x)``: add one item (nothing happens if it's already there)
- ```my_set.remove(x)```: remove x (error if missing)

**Legth, indexing, slicing**:

- ```len(arr)```returns the number of elements in a list / tuple / set.
- _Indexing_: get the value at index i using ```value = arr[i]```, the last element using ```last = arr[-1]```
- _Slicing_: exctract a sub-list with ```sub = arr[start:end]```

**Casting**:

- To change from one array type to another, use explicit casts: ```set(x), list(x), tuple(x)```(**this might change the values in them!**)

````python
# Heres some data about our participant from before. 

locations = ["home", "work", "home", "park", "home", "work"]
moods = [3, 5, 4, 6, 2, 5, 5]
study_days = ("Mon", "Tue", "Wed", "Thu", "Fri")

# --- Working with lists ---

# 1. The participant answered another prompt from the park. Add it to `locations`
#    and print how many entries the list has now.
locations.append("park")
print(len(locations))

# 2. How often was the participant at home?
print(locations.count("home"))

# 3. One "work" entry was a mistake. Remove it and print the list.
#    Which of the two "work" entries was removed?
locations.remove("work")
print(locations)
# -> remove() only deletes the FIRST "work" (index 1), the later one stays

# 4. Sort `moods`, then print it.
moods.sort()
print(moods)

# --- Tuples ---

# 5. At which position is "Wed" in `study_days`?
print(study_days.index("Wed"))
# -> 2, because indexing starts at 0

# --- Casting ---

# 6. Convert `locations` to a set and print it. Compare it to the list:
#    what disappeared, and what changed?
locations_set = set(locations)
print(locations_set)
# -> the duplicates disappeared, and the order changed (sets are unordered)

# 7. How many *different* places did the participant visit? (one line of code)
print(len(set(locations)))

# 8. Convert the set from task 6 back to a list. Is it the same as the original `locations`?
#    Check with ==.
locations_list = list(locations_set)
print(locations_list)
print(locations_list == locations)
# -> False: the duplicates are lost for good (and the order may differ)

# 9. Bonus: Write an if-statement that prints "Duplicates found!" if `moods`
#     contains any value more than once. (hint: compare lengths)
if len(set(moods)) < len(moods):
    print("Duplicates found!")
````

### Dict

Dicts ae used to store **structured data** as **key-value** pairs. **Keys** are usually strings (other types are possible), **values** can be any type (primitives, array-likes, nested dicts...).

#### Syntax:

```python
my_dict_a = {"key": "value"}

my_dict_b = {
    "id": "sphere0",
    "coordinates": (33, 44),
    "properties": {
        "color": "red",
        "radius": 5.55,
        "child_ids": ["rect2", "sphere1"]
    }
}
```

#### Methods

- Lookup: ```value = my_dict["key"]```
- Nested lookup (example from before): ```color = my_dict_b["properties"]["color"]```
- Add new key-value pair: ```my_dict["new_key"] = new_value```
- Get the dict's keys: ```keys = my_dict.keys()```
