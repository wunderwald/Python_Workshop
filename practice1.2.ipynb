### Array-Likes

There are three types that store multiple objects in python:

- list ```[ ... ]```: ordered, mutable
- tuple ```( ... )```: ordered, **immutable**
- set ```{ ... }```: unordered, mutable, **unique elements**

#### Important methods

**List**:

- ```my_list.append(x)```: add x to the end
- ```my_list.remove(x)```: remove the first x (error if x isn't there)
- ```my_list.pop() / my_list.pop(i)	```: remove and return the last item / the item at i
- ```my_list.count(x)```: count how often x appears
- ```my_list.sort(), my_list.reverse()```: sort / reverse the list (*in place*)

**Tuple**:

- ```my_tuple.count(x)```: how often x appears
- ```my_tuple.index(x)```

**Set**

- ```my_set.add(x)``: add one item (nothing happens if it's already there)
- ```my_set.remove(x)```: remove x (error if missing)

**All of them**:

- ```len(arr)```returns the number of elements in a list / tuple / set.
- Lookup: get the value at index i using ```value = arr[i]```

**Casting**:

- To change from one array type to another, use explicit casts: ```set(x), list(x), tuple(x)```(**this might change the values in them!**)

### Dict

Dicts ae used to store **structured data** as **key-value** pairs. **Keys** are usually strings (other types are possible), **values** can be any type (primitives, array-likes, nested dicts...).

#### Syntax:

```
my_dict_a = {"key": "value"}

my_dict_b = {
    "id": "sphere0",
    "coordinates": (33, 44),
    "properties": {
        "color": "red",
        "radius": 5.55,
        "child_ids": ["rect2", "sphere1"]
    }
}
```

#### Methods

- Lookup: ```value = my_dict["key"]```
- Nested lookup (example from before): ```color = my_dict_b["properties"]["color"]```
- Add new key-value pair: ```my_dict["new_key"] = new_value```
- Get the dict's keys: ```keys = my_dict.keys()```
