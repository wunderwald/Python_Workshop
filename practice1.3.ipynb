# Practice 1.3: Loops

### Syntax

**For-loops**:

For-loops run once for each value in a list:

```python
for value in list:
    do_something(value)
```

We can use them to iterate over lists of objects:

```python
for character in ["a", "b", "c"]:
    print(character)

Output:
a
b
c
```

Or we can use them to do someting N times:

```python
for i in range(3):
    print(i)

Output:
0
1
2
```

**The range() function**:

- ```range(x)```-> all integers starting at 0 up until x (x excluded)
- ```range(start, end)```-> all integers starting at 'start' up until 'end' (end excluded)
- For example: ```range(3) => 0, 1, 2; range(4, 7) => 4, 5, 6```

**While loops**:

While loops run as long as a condition is true. 

```python
while (condition):
    do_something()
```

They are executed like this:

- check condition
- if it is False, stop
- if it is True, run the code inside and repeat

**Your tasks ⬇**

````python
stress = [6, 2, 7, 4, 6, 1, 3]
sleep_hours = [
    {"day": "Mon", "hours": 7.0},
    {"day": "Tue", "hours": 5.5},
    {"day": "Wed", "hours": 6.0},
    {"day": "Thu", "hours": 4.5},
    {"day": "Fri", "hours": 8.0},
]

# 1. Use a loop to calculate the average stress level.
#    (hint: start with total = 0 before the loop)
total = 0
for s in stress:
    total += s
average_stress = total / len(stress)
print(average_stress)

# 2. Create a new list `high_stress` that only contains the stress levels of 6 or higher.
#    (hint: start with an empty list and append inside the loop)
high_stress = []
for s in stress:
    if s >= 6:
        high_stress.append(s)
print(high_stress)

# 3. `sleep_hours` is a list of dicts. Loop over it and print one line per day, e.g. "Tue: 5.5 h".
#    Bonus: Add "<- short night" to days with less than 6 hours.
#    (hint: in each iteration, the loop variable is one of the dicts)
for entry in sleep_hours:
    line = f"{entry['day']}: {entry['hours']} h"
    if entry["hours"] < 6:
        line += " <- short night"
    print(line)

# 4. A participant answers 6 prompts per day. Use a while loop to find out
#    after how many days they have answered at least 100 prompts.
answered = 0
days = 0
while answered < 100:
    answered += 6
    days += 1
print(days)
````
