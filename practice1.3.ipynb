# Practice 1.3: Loops

### Syntax

**For-loops**:

For-loops run once for each value in a list:

```python
for value in list:
    do_something(value)
```

We can use them to iterate over lists of objects:

```python
for character in ["a", "b", "c"]:
    print(character)

Output:
a
b
c
```

Or we can use them to do someting N times:

```python
for i in range(3):
    print(i)

Output:
0
1
2
```

**The range() function**:

- ```range(x)```-> all integers starting at 0 up until x (x excluded)
- ```range(start, end)```-> all integers starting at 'start' up until 'end' (end excluded)
- For example: ```range(3) => 0, 1, 2; range(4, 7) => 4, 5, 6```

**While loops**:

While loops run as long as a condition is true. 

```python
while (condition):
    do_something()
```

They are executed like this:

- check condition
- if it is False, stop
- if it is True, run the code inside and repeat

````python
stress = [6, 2, 7, 4, 6, 1, 3]
sleep_hours = [
    {"day": "Mon", "hours": 7.0},
    {"day": "Tue", "hours": 5.5},
    {"day": "Wed", "hours": 6.0},
    {"day": "Thu", "hours": 4.5},
    {"day": "Fri", "hours": 8.0},
]

# 1. Use a loop to calculate the average stress level. -> AVERAGE = TOTAL / NUM_ELEMENTS
#    (hint: start with total = 0 before the loop)

total = 0
num_samples = len(stress)
for sample in stress:
    # total += sample
    total = total + sample

average_stress = total / num_samples
print(f"Average stress: {average_stress}")
# -> Average stress: 4.14

# 2. Create a new list `high_stress` that only contains the stress levels of 6 or higher.
#    (hint: start with an empty list and append inside the loop)
high_stress = []
for sample in stress:
    if sample >= 6:
        high_stress.append(sample)
    else:
        high_stress.append(None)
print(high_stress)

# 3. `sleep_hours` is a list of dicts. Loop over it and print one line per day, e.g. "Tue: 5.5 h".
#    Bonus: Add "<- short night" to days with less than 6 hours.
#    (hint: in each iteration, the loop variable is one of the dicts)
for s in sleep_hours:
    print(f"{s["day"]}: {s["hours"]}")
    if s["hours"] < 6:
        print("⬆ short night")

# 4. A participant answers 6 prompts per day. Use a while loop to find out
#    after how many days they have answered at least 100 prompts.
num_days = 0
total_answered = 0
daily_answered = 6
while total_answered <= 100:
    # simulate one day: update num_days and total_answered
    num_days += 1
    total_answered += daily_answered

print(f"it took {num_days} to answer more than 100 (precisely {total_answered}) prompts.")
````

````python
# Bug 3: These questionnaire items are reverse-coded (1–7 scale), so every score
#        must be flipped: new score = 8 - old score.
# Expected: [6, 2, 1, 7, 4]

scores = [2, 6, 7, 1, 4]
rev_scores = []

for score in scores:
    score = 8 - score
    rev_scores.append(score)

print(rev_scores)
````
