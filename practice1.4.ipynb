### Functions

Functions are reusable pieces of code that may have inputs (arguments) and may have outputs (return values).

#### Syntax

```python
def my_function(arguments):
    do_something()
    return value
```

#### Examples

A function with no inputs and outputs:

```python
def hello_world():
    print("hello world!")

# call it:
hello_world()

Output:
hello world
```

A function with inputs and outputs:

```python
def calc_sum(a, b):
    sum = a + b
    return sum

# call it:
my_sum = calc_sum(1, 2)
print(my_sum)

Output:
3
```

#### Default parameters

We can set default values for parameters. Those will be used if no value is specified.

**Important**: Default parameters must be at the **end** of the parameter list.

```python
def calc_pow(base, exponent=2):
    print(base ** exponent)

calc_pow(3) # prints 9 (3 ** 2)
calc_pow(3, 3) # prints 27 (3 ** 3)
```

### Modules

Modules are python files from which we can import functions or data. 

Given a module called "my_module.py" with two functions ```add(a, d)```, ```sub(a, b)``` and ```mult(a, b)```, these are different ways to import:

#### Specific import

We select the exact things we want to have:

```python
from my_module import add, sub

# we now have access to add and sub
add(1, 2)
```

#### Import all (not recommended)

```python
from my_module import *

# we now have access to add, sub and mult
add(1, 2)
```

#### Full module import

```python
import my_module

# we can use everything inside of my_module but have to specify the module name
my_module.add(1, 2)
```

#### Named module import
```python
# we can give the module an alias
import my_module as mm

# we can use everything inside of my_module using its alias
mm.add(1, 2)
```

### Packages

Packages are collections of modules written by other people. We import them just like our own modules.

- The **standard library** comes with Python and is always available (e.g. ```math```, ```random```, ```statistics```, ```datetime```, ```json```)
- **Third-party packages** have to be installed first (e.g. ```numpy```, ```pandas```, ```matplotlib```). In this workshop, they are already installed for you!

**Syntax**:

```python
import math

# calculate the square root
print(math.sqrt(16))

Output:
4.0
```

```python
from statistics import mean, median

print(mean([3, 5, 4]))
print(median([3, 5, 4, 2, 6]))

Output:
4
4
```

**Installing third-party packages** (in the terminal, not in python):

```
pip install numpy
```

**What is inside a package?**

- ```help(math)``` or ```dir(math)``` lists its contents
- Or: search for the documentation online (e.g. "python statistics module")

**Careful**: never name your own files like a package (e.g. ```random.py```), python will import your file instead!
