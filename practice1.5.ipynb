### List comprehensions

List comprehensions let you use **for loops inside lists**.

Instead of this:

```python
nums = [1, 2, 3]
squares = []
for x in nums:
    squares.append(x ** 2)
```

We can write:

```python
nums = [1, 2, 3]
squares = [x ** 2 for x in nums]
```

### Ternary operator

The ternary operator lets us use an if statement when assigning a value to a variable.

Instead of this:

```python
formal = True
if formal:
    greeting = "Kind regards"
else:
    greeting = "xoxo"
```

We can write:

```python
formal = True
greeting = "Kind regards" if formal else "xoxo"
```

### zip() and enumerate()

**zip()** lets us iterate over two array-likes at once:

```python
a = ('a', 'b', 'c')
b = (1, 2, 3)
for (val_a, val_b) in zip(a, b):
    print(f"{val_a}, {val_b}")

Output:
a, 1
b, 2
c, 3
```

**enumerate()** provides us with **index and value** when iterating:

```python
arr = [":')", "*.*", ">_<"]
for (i, x) in enumerate(arr):
    print(f"{i}: x")

Output:
1: :')
2: *.*
3: >_<
```
