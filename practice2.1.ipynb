### Cleaning data

Real data is rarely clean. Common problems are **duplicated rows** and **wrong data types** (e.g. numbers stored as text).

**Syntax**:

```python
df.duplicated()  # duplicated rows
df.duplicated().sum() # number of duplicated rows
df = df.drop_duplicates() # remove them

df["col"] = df["col"].astype(float) # change the type of a column
df["col"] = df["col"].str.replace("a", "b") # replace text in every row of a column
df["date"] = pd.to_datetime(df["date"]) # convert text to dates
```

**Careful**: most pandas methods return a **new** DataFrame. Don't forget to store the result!

**Your tasks ⬇**

````python
esm = pd.read_csv("data/esm_simulated.csv")

# 1. How many duplicated rows does `esm` have? Remove them.
#    Expected: 3 duplicates, 75 rows remaining

# print(f"esm has {esm.duplicated().sum()} duplicates")
esm = esm.drop_duplicates()
print(len(esm))

# 2. sleep_hours uses German decimal commas ("6,5"), so pandas reads it as text.
#    Replace the commas with dots and convert the column to float. Check with info().
esm["sleep_hours"] = esm["sleep_hours"].str.replace(",",  ".").astype(float)

# 3. Convert the "date" column to dates with pd.to_datetime().
esm["date"] = pd.to_datetime(esm["date"])
print(esm.info())

# 4. Run describe() again. What is the average amount of sleep?
#    Expected: 6.64 (rounded)

print(esm.describe())

# 5. Bonus: Add the participant info to `esm`:
#    esm = esm.merge(participants, on="participant_id")
# ---> What is the average mood per occupation?

esm = esm.merge(participants, on="participant_id")
print(esm.groupby("occupation")["mood"].mean())
````

### Addition: Rows (the index, loc and iloc)

Every DataFrame has an **index**: the names of its rows. By default, rows are simply numbered 0, 1, 2...

```python
df = pd.DataFrame({
    "name": ["Ana", "Ben", "Cem"],
    "age": [31, 27, 45],
})

# iloc: select rows by their POSITION (counting from 0, like lists)
df.iloc[0]  # first row
df.iloc[0:2] # first two rows (2 is excluded)
df.iloc[-1] # last row

# set_index: use a column as row names
df = df.set_index("name")

# loc: select rows (and columns) by their NAME
df.loc["Ben"]  # the row named "Ben"
df.loc["Ben", "age"]  # row "Ben", column "age" => 27
df.loc[["Ana", "Cem"]] # several rows
df.loc[df["age"] > 30] # filters work with loc too

df = df.reset_index()  # turn the index back into a normal column
```

**loc or iloc?**

- `loc` uses the **names** in the index, `iloc` always uses the **position** (0, 1, 2...)
- After filtering or removing rows, the index has gaps (e.g. 0, 2, 5). Then `df.loc[1]` fails, because there is no row named 1 anymore, while `df.iloc[1]` still gives you the second row.
