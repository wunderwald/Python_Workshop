### Docstrings

**Reading docstrings**

`help(my_function)` prints them, and VS Code shows them when you hover over a function name.

**Syntax: short docstring**

For simple functions, one line is enough:

```python
def is_even(n):
    """Return whether n is an even number."""
    return n % 2 == 0
```

**Syntax: full docstring**

For more complex functions, describe the **summary**, the **arguments** (*Parameters*) and the **return value** (*Returns*). This style is used by numpy, pandas, mne and many other scientific packages, so you'll see it in their documentation too:

```python
def greet(name, language="en"):
    """
    Create a greeting for a person.

    Parameters
    ----------
    name : str
        The name of the person.
    language : str, optional
        "en" for English or "de" for German. Default is "en".

    Returns
    -------
    str
        The greeting, e.g. "Hello Ana!".
    """
    if language == "de":
        return f"Hallo {name}!"
    return f"Hello {name}!"
```

- Each argument: `name : type`, then an indented description in the next line
- Arguments with a default value are marked as `optional`

**Syntax: type hints**

Types can also be written directly into the function's first line. They don't change what the function does but they help you (and VS Code) to understand it:

```python
def greet(name: str, language: str = "en") -> str:
```

**Syntax: module docstring**

A module (a `.py` file) can have a docstring too. It goes at the **very top** of the file and summarizes what the module contains:

```python
"""Helper functions for analysing data."""

def mean(values):
    ...
```

`help(my_module)` then shows the module docstring and the docstrings of all its functions.
