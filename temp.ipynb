```python
# basic operations
a = 6
b = 2
print(a + b)    # 8
print(a - b)    # 4
print(a * b)    # 12
print(a / b)    # 3.0

# exponentiation
print(4 ** 2)   # 16

# floor division
print(6.8 // 2) # 3.0

# modulo
print(4 % 3)    # 1
```
