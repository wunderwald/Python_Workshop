````python
for i in range(1, 4):
    square = i ** 2
    cube = i ** 3
    half = i / 2
    stats = {
        'value': i,
        'square': square,
        'cube': cube,
        'half': half
    }
    print(stats)
````

````python
def get_stats(value):
    square = value ** 2
    cube = value ** 3
    half = value / 2
    stats = {
        'value': value,
        'square': square,
        'cube': cube,
        'half': half
    }
    return stats

for i in range(1, 4):
    stats = get_stats(i)
    print(stats)
````

````python
# iterate over two arrays

a = ('a', 'b', 'c')
b = (1, 2, 3)

for i in range(len(a)):
    val_a = a[i]
    val_b = b[i]
    print(f"{val_a}, {val_b}")
````

````python
# zip a and b!

a = ('a', 'b', 'c')
b = (1, 2, 3)

for (val_a, val_b) in zip(a, b):
    print(f"{val_a}, {val_b}")
````

````python
# zip merges two array-likes

a = ('a', 'b', 'c')
b = (1, 2, 3)
print(zip(a, b))
````

````python
def greet(name, language="en"):
    """
    Create a greeting for a person.

    Parameters
    ----------
    name : str
        The name of the person.
    language : str, optional
        "en" for English or "de" for German. Default is "en".

    Returns
    -------
    str
        The greeting, e.g. "Hello Ana!".
    """
    if language == "de":
        return f"Hallo {name}!"
    return f"Hello {name}!"
````
