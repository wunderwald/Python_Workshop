````python
for i in range(1, 4):
    square = i ** 2
    cube = i ** 3
    half = i / 2
    stats = {
        'value': i,
        'square': square,
        'cube': cube,
        'half': half
    }
    print(stats)
````

````python
def get_stats(value):
    square = value ** 2
    cube = value ** 3
    half = value / 2
    stats = {
        'value': value,
        'square': square,
        'cube': cube,
        'half': half
    }
    return stats

for i in range(1, 4):
    stats = get_stats(i)
    print(stats)
````

````python
def greet(name, language="en"):
    """
    Create a greeting for a person.

    Parameters
    ----------
    name : str
        The name of the person.
    language : str, optional
        "en" for English or "de" for German. Default is "en".

    Returns
    -------
    str
        The greeting, e.g. "Hello Ana!".
    """
    if language == "de":
        return f"Hallo {name}!"
    return f"Hello {name}!"
````

````python
from seaborn import lineplot
from eeg_simulation import simulate_eeg

# simulate EEG data and extract one channel 
trials, channels, times = simulate_eeg(n_trials=1)
ch0 = trials[0][0]

# calculate mean
sum = 0
for sample in ch0:
    sum += sample
mean = sum / len(ch0)

# substract mean from samples
ch0_corrected = []
for sample in ch0:
    sample_corrected = sample - mean
    ch0_corrected.append(sample_corrected)
````
